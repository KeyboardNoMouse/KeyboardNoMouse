```python
np.random.seed(42)

n = 180
dates = pd.date_range("2026-01-01", periods=90)

products = [
    "Cappuccino", "Cold Coffee", "Espresso", "Green Tea",
    "Veg Sandwich", "Pasta", "Brownie", "Cheesecake"
]

category_map = {
    "Cappuccino": "Beverage",
    "Cold Coffee": "Beverage",
    "Espresso": "Beverage",
    "Green Tea": "Beverage",
    "Veg Sandwich": "Food",
    "Pasta": "Food",
    "Brownie": "Dessert",
    "Cheesecake": "Dessert"
}

price_map = {
    "Cappuccino": 180,
    "Cold Coffee": 220,
    "Espresso": 140,
    "Green Tea": 130,
    "Veg Sandwich": 150,
    "Pasta": 250,
    "Brownie": 120,
    "Cheesecake": 200
}

chosen_products = np.random.choice(products, n)

df = pd.DataFrame({
    "Date": np.random.choice(dates, n),
    "Product": chosen_products,
    "Region": np.random.choice(
        ["North", "South", "East", "West"], n
    ),
    "Customer_Type": np.random.choice(
        ["Regular", "New", "Member"], n,
        p=[0.45, 0.30, 0.25]
    ),
    "Quantity": np.random.randint(1, 7, n),
    "Discount": np.random.choice(
        [0, 5, 10, 15, 20], n
    ),
    "Rating": np.round(
        np.clip(np.random.normal(4.3, 0.45, n), 2.5, 5.0), 1
    )
})

df["Category"] = df["Product"].map(category_map)
df["Unit_Price"] = df["Product"].map(price_map)

df["Revenue"] = (
    df["Quantity"] * df["Unit_Price"] *
    (1 - df["Discount"] / 100)
).round(2)

cost_ratio = np.random.uniform(0.50, 0.72, n)
df["Cost"] = (df["Revenue"] * cost_ratio).round(2)
df["Profit"] = (df["Revenue"] - df["Cost"]).round(2)

df = df.sort_values("Date").reset_index(drop=True)

df.head(20)
```

```python

plt.figure(figsize=(8, 5))

plt.hist(df['Rating'], bins=8, edgecolor='Black')

plt.title("Customer Rating Histogram")
plt.xlabel("Rating")
plt.ylabel("Number of orders")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(

            df['Discount'],
            df['Profit'],
            alpha = 0.5
)

plt.title('Discount vs profit')
plt.xlabel('Discount')
plt.ylabel('Profit')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data = df,
    x = "Revenue", bins = 15, kde = True
)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.violinplot(
    data = df,
    x = "Category",
    y = "Revenue"
)

plt.title("Revenue vs category")
plt.xlabel("category")
plt.ylabel("Revenue")

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="Revenue",
    y="Profit",
    hue = "Category",
    size = "Quantity",
    alpha =0.8
)

plt.title("Revenue vs Profit")
plt.show()
```

```python
numeric_column = [
    "Quantity",
    "Unit_Price",
    "Revenue",
    "Discount",
    "Cost",
    "Profit",
    "Rating"
]

correlation = df[numeric_column].corr()


correlation
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    correlation,
    annot = True,
    fmt = ".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.pairplot(
    df[
        [
            "Quantity",
            "Revenue",
            "Profit",
            "Rating",
            "Category"
        ]
    ],
    hue = "Category"
)

plt.show()
```
